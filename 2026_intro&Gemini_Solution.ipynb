**Conditional:** Python offers if, elif, and else for performing different operations depending on certain conditions.

> Action: Change the number to different numbers and observe the output
```python
number = 15
```

````python
# conditional

# Initialize a variable with the value of 15
number = 15

# Check if the number is less than 10
if number < 10:
    # If the condition is true, print this message
    print("The number is less than 10.")
# If the first condition is not met, check if the number is less than 20
elif number < 20:
    # If the second condition is true, print this message
    print("The number is between 10 and 19.")
# If none of the above conditions are met
else:
    # Print this message
    print("The number is 20 or greater.")
````

**Loops:** in many cases, we need to perform the same operation multiple times. For such scenarios, we can use for or while loops, depending on which works better.

There are a few types of loops in python. Here are the two:

## for loops:
For loops will go through a list or a range of numbers in a list and run the code in the loop until it reaches the end of the list.
Example:
```python
for i in range (1,10):
  print (i)

```

## while loops:
While loops with loop the code inside it until a condition is met.
Caution: If you are not carful you can make an ifinite loop...

```python
counter = 0
while n is not True:
  counter += counter + 1
  if counter > 10:
    n = True
```

> Action: Run the two examples below and observe the outputs


````python
# for loops: these are great when the number of iteration is known.

# Define a list of fruits
fruits = ["apple", "banana", "cherry"]

# Loop through each item in the fruits list
for fruit in fruits:
    # Print the current fruit in the loop
    print(fruit)
````

````python
# while loops: these are great when the number of iterations needed is not known

# Initialize a variable with a value of 1
n = 1

# Continue looping as long as n is less than 100
while n < 100:
    # Double the value of n
    n *= 2

# Once the loop ends, print the final value of n
print(n)
````

**Plotting**: To plot using Python, we will use ```matplotlib``` library. This library has syntax for plotting many types of figures. You can find examples on the website: https://matplotlib.org/stable/

> **Action:** Run the cell below to plot a line graph
